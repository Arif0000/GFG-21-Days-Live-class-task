```python
import sqlite3
import pandas as pd


conn = sqlite3.connect("company.db")


def text_to_sql(user_query):

    user_query = user_query.lower()


    if "all employees" in user_query:
        return "SELECT * FROM employees"


    elif "names" in user_query:
        return "SELECT name FROM employees"


    elif "salary" in user_query and "department" in user_query:
        return """
        SELECT e.name, e.salary, d.dept_name
        FROM employees e
        JOIN departments d
        ON e.dept_id = d.dept_id
        """


    elif "average salary" in user_query:
        return """
        SELECT AVG(salary) as avg_salary
        FROM employees
        """


    elif "engineering" in user_query:
        return """
        SELECT e.name
        FROM employees e
        JOIN departments d
        ON e.dept_id = d.dept_id
        WHERE d.dept_name = 'Engineering'
        """

    else:
        return "SELECT * FROM employees"



def run_query(user_query):
    sql = text_to_sql(user_query)
    print("\nGenerated SQL:\n", sql)

    df = pd.read_sql_query(sql, conn)
    print("\nResult:\n", df)



run_query("Show all employees")
run_query("Show names of employees")
run_query("Show salary with department")
run_query("What is average salary?")
run_query("Show employees in Engineering")


conn.close()
```
